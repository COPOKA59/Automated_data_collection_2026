### Задание 1
Посмотреть документацию к API GitHub, разобраться как вывести список репозиториев для конкретного пользователя,
сохранить JSON-вывод в файле *.json.


```python
USERNAME = "COPOKA59"
```

```python
url = f"https://api.github.com/users/{USERNAME}/repos"
headers = {
    "Accept": "application/vnd.github+json"
}
if TOKEN:
    headers["Authorization"] = f"Bearer {TOKEN}"

response = requests.get(url, headers=headers)

if response.status_code == 200:
    repos = response.json()
    repo_names = [repo["name"] for repo in repos]

    print(f"Всего репозиториев: {len(repo_names)}\n")
    for i, name in enumerate(repo_names, start=1):
        print(f"{i:>2}. {name}")
else:
    print(f"Ошибка: {response.status_code} — {response.text}")
```

```python
# Сохраняем JSON в файл
if response.status_code == 200:
    with open("github_repos.json", "w", encoding="utf-8") as f:
        json.dump(repos, f, indent=4, ensure_ascii=False)
    print("Файл github_repos.json сохранён.")
else:
    print("Сначала исправьте ошибку в предыдущей ячейке.")
```

Используем NewsAPI.org для получения топ-новостей.
- Авторизация: API-ключ передаётся как query-параметр `API_KEY`.
- Эндпоинт: `/v2/everything`

```python
BASE_URL = "https://newsapi.org/v2"
```

```python
url = f"{BASE_URL}/everything"
# Ищем новости по параметрам:
params = {
    "q": "Python",
    "language": "ru",
    "sortBy": "publishedAt",
    "pageSize": 10,
    "apiKey": API_KEY
}

response = requests.get(url, params=params)
search_data = response.json()

if search_data.get("status") == "ok":
    print(f"Найдено статей по запросу 'Python': {search_data['totalResults']}")
    print()
    for i, article in enumerate(search_data["articles"], 1):
        print(f"{i}. {article['title']}")
        print(f"   {article['url']}\n")

    with open("news_search_python.json", "w", encoding="utf-8") as f:
        json.dump(search_data, f, indent=4, ensure_ascii=False)
    print("Файл news_search_python.json сохранён.")
else:
    print(f"Ошибка: {search_data.get('message')}")
```
